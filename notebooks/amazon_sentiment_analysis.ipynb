```python
import pandas as pd
from textblob import TextBlob

# 1. Load the raw dataset
url = "https://raw.githubusercontent.com/pycaret/pycaret/master/datasets/amazon.csv"
df = pd.read_csv(url)

# 2. Inspect & Clean Text
review_column = 'text' if 'text' in df.columns else df.columns[0]
df['cleaned_review'] = df[review_column].astype(str).str.lower().str.strip()

# 3. Perform Sentiment Classification using TextBlob
def get_sentiment(text):
    analysis = TextBlob(str(text))
    if analysis.sentiment.polarity > 0.05:
        return 'Positive'
    elif analysis.sentiment.polarity < -0.05:
        return 'Negative'
    else:
        return 'Neutral'

df['sentiment'] = df['cleaned_review'].apply(get_sentiment)

# 4. Create dummy categorical columns for Power BI slicing if not present
if 'rating' not in df.columns:
    df['rating'] = 5 # Default fallback rating placeholder

# 5. Export final cleaned data for Power BI
df.to_csv('cleaned_amazon_reviews.csv', index=False)
print("Data processing complete. 'cleaned_amazon_reviews.csv' is ready for Power BI!")
```
